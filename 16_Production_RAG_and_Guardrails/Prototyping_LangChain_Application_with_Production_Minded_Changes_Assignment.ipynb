#### Production Caching Architecture

Our LLMOps library implements sophisticated caching at multiple levels:

**Embedding Caching:**
The process of embedding is typically very time consuming and expensive:

1. Send text to OpenAI API endpoint
2. Wait for processing  
3. Receive response
4. Pay for API call

This occurs *every single time* a document gets converted into a vector representation.

**Our Caching Solution:**
1. Check local cache for previously computed embeddings
2. If found: Return cached vector (instant, free)
3. If not found: Call OpenAI API, store result in cache
4. Return vector representation

**LLM Response Caching:**
Similarly, we cache LLM responses to avoid redundant API calls for identical prompts.

**Benefits:**
- ⚡ Faster response times (cache hits are instant)
- 💰 Reduced API costs (no duplicate calls)  
- 🔄 Consistent results for identical inputs
- 📈 Better scalability

Our ProductionRAGChain automatically handles all this caching behind the scenes!

##### ❓ Question #1: Production Caching Analysis

What are some limitations you can see with this caching approach? When is this most/least useful for production systems? 

Consider:
- **Memory vs Disk caching trade-offs**
- **Cache invalidation strategies** 
- **Concurrent access patterns**
- **Cache size management**
- **Cold start scenarios**

> NOTE: There is no single correct answer here! Discuss the trade-offs with your group.

##### ✅ Answer

**Limitations of Current Caching Approach:**

1. **Memory vs Disk Trade-offs:**
   - **Memory caching (InMemoryCache)**: Fast but volatile - lost on restart, limited by RAM
   - **Disk caching (SQLiteCache)**: Persistent but slower I/O, better for production
   - **Embedding cache (LocalFileStore)**: Disk-based, persistent, but needs file system management

2. **Cache Invalidation Challenges:**
   - No automatic expiration - stale data can persist indefinitely
   - No versioning - if document changes, cached embeddings become outdated
   - No invalidation strategy for related data (e.g., if PDF is updated)

3. **Concurrent Access Patterns:**
   - In-memory cache: Not thread-safe for concurrent writes
   - SQLite cache: Better for concurrent reads but can have write contention
   - File-based embedding cache: File system locks can cause bottlenecks

4. **Cache Size Management:**
   - No size limits - cache can grow unbounded
   - No LRU/LFU eviction policies
   - Disk space can fill up with embedding cache files
   - Memory can be exhausted with large in-memory caches

5. **Cold Start Scenarios:**
   - First request always slow (no cache)
   - No pre-warming strategies
   - Cache misses penalize first-time queries significantly

**When Most Useful:**
- **Development/Prototyping**: Fast iteration with repeated queries
- **Stable Document Sets**: When documents rarely change
- **Repeated Queries**: High query overlap scenarios
- **Cost Optimization**: Reducing API calls for expensive embeddings

**When Least Useful:**
- **Frequently Changing Documents**: Cache invalidation becomes complex
- **Low Query Overlap**: If queries are always unique, caching provides minimal benefit
- **High Concurrency**: Without proper locking, cache can become a bottleneck
- **Real-time Requirements**: Cache lookups add latency even when fast

Create a simple experiment that tests our production caching system:

1. **Test embedding cache performance**: Try embedding the same text multiple times
2. **Test LLM cache performance**: Ask the same question multiple times

````python
### Cache Performance Testing Experiment

import time
import statistics
from langchain_core.messages import HumanMessage

print("🧪 Cache Performance Testing Experiment")
print("=" * 60)

# Test 1: Embedding Cache Performance
print("\n1️⃣ Testing Embedding Cache Performance")
print("-" * 60)

test_texts = [
    "What is the Direct Loan Program?",
    "How do I apply for student loans?",
    "What are the repayment options available?",
    "What is the Direct Loan Program?",  # Duplicate to test cache
    "How do I apply for student loans?",  # Duplicate to test cache
]

embedding_times = []
cache_hits = 0
cache_misses = 0

for i, text in enumerate(test_texts):
    start_time = time.time()
    # Embed the text using our RAG chain's embedding system
    embeddings = rag_chain.cached_embeddings.get_embeddings()
    result = embeddings.embed_query(text)
    elapsed = time.time() - start_time
    embedding_times.append(elapsed)
    
    if i < len(test_texts) - 2 and text in test_texts[:i]:
        cache_hits += 1
        print(f"  ⚡ Cache HIT: '{text[:50]}...' - {elapsed:.4f}s")
    else:
        cache_misses += 1
        print(f"  🔄 Cache MISS: '{text[:50]}...' - {elapsed:.4f}s")

first_call_avg = statistics.mean(embedding_times[:3])  # First 3 unique
cached_call_avg = statistics.mean(embedding_times[3:])  # Last 2 (cached)
speedup = first_call_avg / cached_call_avg if cached_call_avg > 0 else float('inf')

print(f"\n📊 Embedding Cache Results:")
print(f"  - First calls (cache miss) avg: {first_call_avg:.4f}s")
print(f"  - Cached calls (cache hit) avg: {cached_call_avg:.4f}s")
print(f"  - Speedup: {speedup:.2f}x faster")
print(f"  - Cache hit rate: {cache_hits}/{len(test_texts)} ({100*cache_hits/len(test_texts):.1f}%)")

# Test 2: LLM Cache Performance
print("\n2️⃣ Testing LLM Cache Performance")
print("-" * 60)

llm_queries = [
    "What is this document about?",
    "Explain the loan repayment process.",
    "What is this document about?",  # Duplicate
    "Explain the loan repayment process.",  # Duplicate
]

llm_times = []
llm_cache_hits = 0
llm_cache_misses = 0

for i, query in enumerate(llm_queries):
    start_time = time.time()
    response = rag_chain.invoke(query)
    elapsed = time.time() - start_time
    llm_times.append(elapsed)
    
    if i < len(llm_queries) - 2 and query in llm_queries[:i]:
        llm_cache_hits += 1
        print(f"  ⚡ Cache HIT: '{query[:50]}...' - {elapsed:.2f}s")
        print(f"     Response preview: {response.content[:80]}...")
    else:
        llm_cache_misses += 1
        print(f"  🔄 Cache MISS: '{query[:50]}...' - {elapsed:.2f}s")
        print(f"     Response preview: {response.content[:80]}...")

first_llm_avg = statistics.mean(llm_times[:2])  # First 2 unique
cached_llm_avg = statistics.mean(llm_times[2:])  # Last 2 (cached)
llm_speedup = first_llm_avg / cached_llm_avg if cached_llm_avg > 0 else float('inf')

print(f"\n📊 LLM Cache Results:")
print(f"  - First calls (cache miss) avg: {first_llm_avg:.2f}s")
print(f"  - Cached calls (cache hit) avg: {cached_llm_avg:.2f}s")
print(f"  - Speedup: {llm_speedup:.2f}x faster")
print(f"  - Cache hit rate: {llm_cache_hits}/{len(llm_queries)} ({100*llm_cache_hits/len(llm_queries):.1f}%)")

# Test 3: Combined Cache Performance (RAG Chain)
print("\n3️⃣ Testing Combined RAG Chain Cache Performance")
print("-" * 60)

rag_queries = [
    "What are the eligibility requirements?",
    "What are the eligibility requirements?",  # Exact duplicate
    "What are the eligibility requirements for Direct Loans?",  # Similar but different
]

rag_times = []
for i, query in enumerate(rag_queries):
    start_time = time.time()
    response = rag_chain.invoke(query)
    elapsed = time.time() - start_time
    rag_times.append(elapsed)
    
    if i == 1:
        print(f"  ⚡ Exact duplicate (full cache hit): {elapsed:.2f}s")
    elif i == 0:
        print(f"  🔄 First query (full cache miss): {elapsed:.2f}s")
    else:
        print(f"  🔄 Similar query (partial cache hit): {elapsed:.2f}s")

print(f"\n📊 Combined RAG Cache Results:")
print(f"  - First query: {rag_times[0]:.2f}s")
print(f"  - Exact duplicate: {rag_times[1]:.2f}s ({rag_times[0]/rag_times[1]:.2f}x faster)")
print(f"  - Similar query: {rag_times[2]:.2f}s")

# Summary
print("\n" + "=" * 60)
print("🎯 Cache Performance Summary")
print("=" * 60)
print(f"Embedding Cache: {speedup:.2f}x speedup")
print(f"LLM Cache: {llm_speedup:.2f}x speedup")
print(f"Combined RAG: {rag_times[0]/rag_times[1]:.2f}x speedup for exact duplicates")
print("\n✅ Cache performance testing complete!")
````

##### ✅ Answer

**Note:** The notebook mentions a "Helpfulness Agent" with response evaluation capabilities, but only the Simple Agent is currently implemented in the codebase. The comparison below discusses the theoretical helpfulness agent architecture as described in the notebook materials.

**1. When to Choose Each Agent Type:**

**Simple Agent Advantages:**
- ✅ **Lower Latency**: Direct response without evaluation loops
- ✅ **Lower Cost**: Single LLM call per query (plus tools)
- ✅ **Faster Response**: No iterative refinement overhead
- ✅ **Simpler Architecture**: Easier to debug and monitor
- ✅ **Better for High-Volume**: Can handle more concurrent requests

**Simple Agent Disadvantages:**
- ❌ **No Quality Assurance**: Responses aren't validated for helpfulness
- ❌ **Potential for Poor Responses**: May return incomplete or irrelevant answers
- ❌ **No Self-Reflection**: Cannot refine or improve its own responses

**Helpfulness Agent Advantages:**
- ✅ **Quality Assurance**: Built-in evaluation ensures responses meet quality standards
- ✅ **Self-Reflection**: Can identify and fix issues in its own responses
- ✅ **Better User Experience**: More likely to provide helpful, complete answers
- ✅ **Production Safety**: Reduces risk of poor quality responses reaching users

**Helpfulness Agent Disadvantages:**
- ❌ **Higher Latency**: Multiple LLM calls (initial + evaluation + potential refinement)
- ❌ **Higher Cost**: 2-3x more API calls per query
- ❌ **More Complex**: Harder to debug and monitor
- ❌ **Slower Throughput**: Lower request handling capacity

**2. Production Considerations:**

**Latency Impact:**
- Helpfulness check adds 1-2 seconds per query (additional LLM call)
- For high-traffic systems, this can significantly impact user experience
- Consider async evaluation or parallel processing to mitigate

**Cost Implications:**
- Helpfulness agent: ~2-3x API costs (original + evaluation + potential refinement)
- For 1M queries/month: Simple agent ~$1000, Helpfulness agent ~$2500-3000
- ROI depends on quality improvement and reduction in support tickets

**Monitoring Strategy:**
- Track: response times, cost per query, evaluation pass/fail rates
- Alert on: high failure rates, latency spikes, cost anomalies
- Monitor: tool selection patterns, cache hit rates, error rates

**3. Scalability Questions:**

**Concurrent Load Performance:**
- Simple Agent: Better - fewer LLM calls, faster responses, higher throughput
- Helpfulness Agent: Limited by evaluation overhead, may need async processing

**Caching Strategies:**
- **Simple Agent**: Cache tool results, LLM responses, embeddings
- **Helpfulness Agent**: Cache evaluations for similar queries, cache tool results
- Both benefit from semantic caching for similar (not identical) queries

**Rate Limiting & Circuit Breakers:**
- Implement rate limits per user/IP to prevent abuse
- Circuit breakers for tool failures (e.g., if Tavily is down)
- Exponential backoff for retries
- Graceful degradation (fallback to simple agent if helpfulness agent fails)

**Recommendation:**
- Use **Simple Agent** for: high-volume, low-latency requirements, cost-sensitive applications
- Use **Helpfulness Agent** for: customer-facing applications, quality-critical use cases, low-volume/high-value queries

### 🛡️ What are Guardrails?

Guardrails are specialized validation systems that help "catch" when LLM interactions go outside desired parameters. They operate both **pre-generation** (input validation) and **post-generation** (output validation) to ensure safe, compliant, and on-topic responses.

**Key Categories:**
- **Topic Restriction**: Ensure conversations stay on-topic
- **PII Protection**: Detect and redact sensitive information  
- **Content Moderation**: Filter inappropriate language/content
- **Factuality Checks**: Validate responses against source material
- **Jailbreak Detection**: Prevent adversarial prompt attacks
- **Competitor Monitoring**: Avoid mentioning competitors

### Production Benefits of Guardrails

**🏢 Enterprise Requirements:**
- **Compliance**: Meet regulatory requirements for data protection
- **Brand Safety**: Maintain consistent, appropriate communication tone
- **Risk Mitigation**: Reduce liability from inappropriate AI responses
- **Quality Assurance**: Ensure factual accuracy and relevance

- **Layered Defense**: Multiple validation stages for robust protection
- **Selective Enforcement**: Different guards for different use cases

````python
### Implementation: Building Production-Safe LangGraph Agent with Guardrails

import logging
import warnings
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langchain_core.messages import BaseMessage, HumanMessage, AIMessage
from typing_extensions import TypedDict, Annotated
from langgraph.graph.message import add_messages

# Set up logging - suppress expected validation errors
logging.basicConfig(level=logging.WARNING)  # Only show warnings and errors
logger = logging.getLogger(__name__)

# Suppress harmless UserWarnings from guardrails about event loops
warnings.filterwarnings("ignore", category=UserWarning, module="guardrails.validator_service")

# Import guardrails utilities
from langgraph_agent_lib.guardrails import (
    create_guardrails_guard,
    create_factuality_guard,
    create_guardrails_node,
    validate_input,
    validate_output,
    GuardrailsState
)

print("🛡️ Setting up Production-Safe LangGraph Agent with Guardrails...")
print("=" * 70)

# Create input guard (validates user queries)
print("\n1️⃣ Creating Input Guard...")
input_guard = create_guardrails_guard(
    valid_topics=["student loans", "financial aid", "education financing", "loan repayment", "Direct Loan Program"],
    invalid_topics=["investment advice", "cryptocurrency", "gambling", "politics", "unrelated topics"],
    enable_jailbreak_detection=True,
    enable_pii_protection=True,
    enable_profanity_check=True,
    enable_competitor_check=False
)
print("✅ Input guard configured (topic restriction, jailbreak detection, PII protection)")

# Create output guard (validates agent responses)
print("\n2️⃣ Creating Output Guard...")
output_guard = create_guardrails_guard(
    valid_topics=["student loans", "financial aid", "education financing", "loan repayment"],
    invalid_topics=["investment advice", "cryptocurrency", "gambling", "politics"],
    enable_profanity_check=True,
    enable_pii_protection=True
)
print("✅ Output guard configured (content moderation, PII protection)")

# Create factuality guard for RAG responses
print("\n3️⃣ Creating Factuality Guard...")
try:
    factuality_guard = create_factuality_guard(eval_model="gpt-4.1-mini", on_prompt=True)
    print("✅ Factuality guard configured (RAG response validation)")
except Exception as e:
    print(f"⚠ Factuality guard not available: {e}")
    factuality_guard = None

print("\n✅ All guardrails configured successfully!")
````

````python
### Creating Guardrails-Enabled Agent Function

def create_guardrails_agent(
    model_name: str = "gpt-4.1-mini",
    temperature: float = 0.1,
    rag_chain=None,
    input_guard=None,
    output_guard=None,
    strict_mode: bool = True
):
    """Create a LangGraph agent with integrated guardrails.
    
    Args:
        model_name: OpenAI model name
        temperature: Model temperature
        rag_chain: Optional RAG chain for document retrieval
        input_guard: Guard for input validation
        output_guard: Guard for output validation
        strict_mode: If True, raise exceptions on validation failure
        
    Returns:
        Compiled LangGraph agent with guardrails
    """
    from langgraph_agent_lib.agents import get_default_tools, create_rag_tool
    from langgraph_agent_lib.models import get_openai_model
    
    # Get tools
    tools = get_default_tools(rag_chain)
    
    # Get model and bind tools
    model = get_openai_model(model_name=model_name, temperature=temperature)
    model_with_tools = model.bind_tools(tools)
    
    # Create guardrails nodes
    input_guard_node = create_guardrails_node(
        input_guard=input_guard,
        output_guard=None,
        strict_mode=strict_mode
    )
    
    output_guard_node = create_guardrails_node(
        input_guard=None,
        output_guard=output_guard,
        strict_mode=strict_mode
    )
    
    # Agent state with guardrails
    class GuardedAgentState(GuardrailsState):
        pass
    
    def call_model(state: GuardedAgentState) -> dict:
        """Invoke the model with messages."""
        messages = state["messages"]
        response = model_with_tools.invoke(messages)
        return {"messages": [response]}
    
    def should_continue(state: GuardedAgentState):
        """Route to tools if the last message has tool calls."""
        last_message = state["messages"][-1]
        if getattr(last_message, "tool_calls", None):
            return "action"
        return "validate_output"
    
    def route_after_input_validation(state: GuardedAgentState):
        """Route after input validation."""
        validation_results = state.get("validation_results", [])
        # Check if any validation failed
        if validation_results:
            # validation_results is a list of dicts
            for result in validation_results:
                if isinstance(result, dict) and not result.get("passed", True):
                    return "reject_input"
        return "agent"
    
    def reject_input_handler(state: GuardedAgentState) -> dict:
        """Handle rejected inputs."""
        error_msg = AIMessage(
            content="I'm sorry, but I can only help with questions related to student loans, financial aid, and education financing. Please ask a question within these topics."
        )
        return {"messages": [error_msg]}
    
    # Build graph
    graph = StateGraph(GuardedAgentState)
    tool_node = ToolNode(tools)
    
    # Add nodes
    graph.add_node("validate_input", input_guard_node)
    graph.add_node("agent", call_model)
    graph.add_node("action", tool_node)
    graph.add_node("validate_output", output_guard_node)
    graph.add_node("reject_input", reject_input_handler)
    
    # Set entry point
    graph.set_entry_point("validate_input")
    
    # Add edges
    graph.add_conditional_edges(
        "validate_input",
        route_after_input_validation,
        {
            "agent": "agent",
            "reject_input": "reject_input"
        }
    )
    
    graph.add_conditional_edges(
        "agent",
        should_continue,
        {
            "action": "action",
            "validate_output": "validate_output"
        }
    )
    
    graph.add_edge("action", "agent")
    graph.add_edge("validate_output", END)
    graph.add_edge("reject_input", END)
    
    return graph.compile()


# Create the guardrails-enabled agent
print("\n4️⃣ Creating Guardrails-Enabled Agent...")
try:
    guardrails_agent = create_guardrails_agent(
        model_name="gpt-4.1-mini",
        temperature=0.1,
        rag_chain=rag_chain,
        input_guard=input_guard,
        output_guard=output_guard,
        strict_mode=False  # Use non-strict mode for graceful error handling
    )
    print("✅ Guardrails-enabled agent created successfully!")
    print("   - Input validation: Active")
    print("   - Output validation: Active")
    print("   - Error handling: Graceful")
except Exception as e:
    print(f"❌ Error creating guardrails agent: {e}")
    guardrails_agent = None
````
