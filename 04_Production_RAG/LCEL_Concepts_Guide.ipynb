# LangChain and LCEL Concepts Guide

This notebook provides a comprehensive guide to understanding LangChain Expression Language (LCEL) and how to build powerful RAG applications with it.

## What is LCEL?

LangChain Expression Language (LCEL) is a declarative way to compose chains of LangChain components. It makes building complex applications simple and intuitive.

### Key Benefits:
- **Composability**: Easy to combine different components
- **Streaming**: Built-in support for streaming responses
- **Parallelization**: Automatic batching and parallel processing
- **Debugging**: Easy to inspect intermediate outputs
- **Type Safety**: Better error handling and type checking


```python
# Let's start by setting up our environment
import os
from langchain_ollama import ChatOllama, OllamaEmbeddings
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import HumanMessage, SystemMessage

# Initialize our models
chat_model = ChatOllama(model="gpt-oss:20b", temperature=0.7)
embedding_model = OllamaEmbeddings(model="embeddinggemma:latest")

print("✅ Models initialized successfully!")
```

```python
# Example 1: Simple prompt + model + parser chain
prompt = ChatPromptTemplate.from_template("Explain {topic} in simple terms.")
output_parser = StrOutputParser()

# Create a chain using the | operator
simple_chain = prompt | chat_model | output_parser

# Test the chain
result = simple_chain.invoke({"topic": "quantum computing"})
print("Simple Chain Result:")
print(result)
```
